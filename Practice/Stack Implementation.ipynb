```python
# add delimiter match code

def delimiter_match(expr):
    stack = []

    pairs = {
        ')': '(',
        ']': '[',
        '}': '{'
    }

    for ch in expr:

        if ch in '([{':
            stack.append(ch)

        elif ch in ')]}':

            if not stack:
                return False

            if stack.pop() != pairs[ch]:
                return False

    return len(stack) == 0
```

```python
# test of delimiter match code

print(delimiter_match("(a+b)*[c-d]"))      
print(delimiter_match("([{}])"))           
print(delimiter_match("(a+b]*{c-d}"))      
print(delimiter_match("((a+b)"))          
```

```python
# postfix code evaluation

def evaluate_postfix(expr):

    stack = []

    for ch in expr:

        if ch.isdigit():
            stack.append(int(ch))

        elif ch in "+-*/%":

            op2 = stack.pop()
            op1 = stack.pop()

            if ch == '+':
                stack.append(op1 + op2)

            elif ch == '-':
                stack.append(op1 - op2)

            elif ch == '*':
                stack.append(op1 * op2)

            elif ch == '/':
                stack.append(op1 / op2)

            elif ch == '%':
                stack.append(op1 % op2)

    return stack.pop()
```

```python
# test of postfix code evaluation

expr = "1 3 5 * +"

print(evaluate_postfix(expr))
```

```python
# code for reversing a string using recursion

def rev():
    ch = input("Enter alphabet: ")

    if ch.lower() == 'z':
        return

    rev()

    print(ch)

if __name__ == '__main__':
    rev()
```

```python
# code for reversing a string using stack

def reverse_text(text):
    stack = []

    for ch in text:
        stack.append(ch)

    result = ''

    while stack:
        result += stack.pop()

    return result

print(reverse_text("HELLO"))
```

```python
# another code for reversing a string using stack

def reverse_string(text):
    stack = list(text)

    result = ""

    while stack:
        result += stack.pop()

    return result

print(reverse_string("HELLO"))
```
