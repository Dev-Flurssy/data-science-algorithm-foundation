```python
# code for enqueuing an element with given size

def enqueue(q, el, max_size):
    if len(q) < max_size:
        q.append(el)
        return q
    return 'Queue Full'

print(enqueue([1, 7, 9], 3, 5))
```

```python
# code for dequeuing an element

def dequeue(q):
    if len(q) > 0:
        return q.pop(0)

    return 'Queue Empty'

q = [1, 7, 9]

q = dequeue(q)
print(q)
```

```python
# class queue implementation

class Queue:

    def __init__(self, size):
        self.items = []
        self.size = size

    def is_Empty(self):
        return len(self.items) == 0

    def is_Full(self):
        return len(self.items) >= self.size

    def enqueue(self, el):
        if self.is_Full():
            return "Queue Full"

        self.items.append(el)

    def dequeue(self):
        if self.is_Empty():
            return "Queue Empty"

        return self.items.pop(0)

    def display(self):
        if self.is_Empty():
            return "Queue Empty"

        return self.items
```

```python
# test of code for class queue implementation

q = Queue(5)

print("Queue Empty?:", q.is_Empty())
print()

q.enqueue(23)
q.enqueue(9)
q.enqueue(41)
q.enqueue(0)

print("Queue:", q.display())
print("Queue Full?:", q.is_Full())
print()

q.enqueue(17)
q.enqueue(82)  
q.enqueue(35)  

print("Queue:", q.display())
print("Queue Full?:", q.is_Full())
print()

print("Dequeued:", q.dequeue())
print("Dequeued:", q.dequeue())

print("Queue:", q.display())
```

```python
# another code for dequeuing using a dictionary

def dequeue(q):
    if not q:
        return 'Queue Empty'
    front = min(q.keys())
    removed = q.pop(front)
    
    print('Dequeued:', removed)
    return q

q = {0: 1, 1:9, 2: 7, 3: 0, 4:5}
print(dequeue(q))
```

```python
# anodther code for displaying queue using a dictionary

def display(q):
    if not q:
        print('Queue Empty')
        return

    print('Queue Elements:')
    for key, value in q.items():
        print(key, '->', value)
    return

q = {0: 1, 1:9, 2: 7, 3: 0, 4:5}
print(display(q))
```

```python
# code for class queue implementation using a dictionary

class Menu:

    def __init__(self, size):
        self.items = {}
        self.size = size

    def is_Empty(self):
        return len(self.items) == 0

    def is_Full(self):
        return len(self.items) >= self.size

    def enqueue(self, el):
        if self.is_Full():
            return "Queue Full"

        next_key = max(self.items.keys(), default=-1) + 1
        self.items[next_key] = el

        return self.items

    def dequeue(self):
        if self.is_Empty():
            return "Queue Empty"

        front = min(self.items.keys())
        return self.items.pop(front)

    def display(self):
        if self.is_Empty():
            return "Queue Empty"

        return self.items
```

```python
# code for creating queue implementation using queue module

from queue import Queue

q = Queue(maxsize=5)


def enqueue(q):
    if q.full():
        return 'Queue Full'

    data = int(input('Enter Queue Number: '))
    q.put(data)
    return f'{data} inserted Successfully!'


def dequeue(q):
    if q.empty():
        return 'Queue Empty'

    data = q.get()
    return f'{data} deleted Successfully!'


def display(q):
    if q.empty():
        return 'Queue Empty'

    return list(q.queue)


def menu():
    print('1. Enqueue')
    print('2. Dequeue')
    print('3. Display')
    print('4. Exit')

    choice = int(input('Please enter number 1 - 4: '))

    if choice == 1:
        print(enqueue(q))

    elif choice == 2:
        print(dequeue(q))

    elif choice == 3:
        print(display(q))

    elif choice == 4:
        return "Program ended."

    else:
        print("Invalid choice.")


print(menu())
```

```python
# circular Queue class implementation

class MyCircularQueue:
    def __init__(self, size):
        self.size = size
        self.queue = [None] * size
        self.front = -1
        self.rear = -1

    def enqueue(self):
        data = int(input("Enter a number to insert into the Queue: "))

        # Queue Full
        if (self.rear + 1) % self.size == self.front:
            print("Circular Queue is Full")

        # First Element
        elif self.front == -1:
            self.front = 0
            self.rear = 0
            self.queue[self.rear] = data

        # Normal Insertion
        else:
            self.rear = (self.rear + 1) % self.size
            self.queue[self.rear] = data

    def dequeue(self):
        # Queue Empty
        if self.front == -1:
            print("Circular Queue is Empty")

        # Only One Element
        elif self.front == self.rear:
            print("Deleted:", self.queue[self.front])
            self.queue[self.front] = None
            self.front = -1
            self.rear = -1

        # More Than One Element
        else:
            print("Deleted:", self.queue[self.front])
            self.queue[self.front] = None
            self.front = (self.front + 1) % self.size

    def display(self):
        print("Front Index:", self.front)
        print("Rear Index :", self.rear)

        if self.front == -1:
            print("Queue Empty")
        else:
            print("Queue:", self.queue)


def menu():
    print("\n1. Enqueue")
    print("2. Dequeue")
    print("3. Display")
    print("4. Exit")

    return int(input("Enter choice: "))


# Main Program
size = int(input("Enter Queue Size: "))
cq = MyCircularQueue(size)

while True:
    choice = menu()

    if choice == 1:
        cq.enqueue()
        cq.display()

    elif choice == 2:
        cq.dequeue()
        cq.display()

    elif choice == 3:
        cq.display()

    elif choice == 4:
        print("Exit")
        break

    else:
        print("Invalid Choice")
```

```python
# double ended queue insertion implementation

from collections import deque

def insert(q):
    data = int(input('Enter Number: '))
    side = input('F = Front, R = Rear: ').upper()

    if side == 'F':
        q.appendleft(data)

    elif side == 'R':
        q.append(data)

    else:
        print('Invalid Choice')

q = deque([20, 30])
insert(q)

print(list(q))
```
