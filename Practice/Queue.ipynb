```python
# creating a queue using list

def create(*s):
    q = []
    q.extend(s)
    return q


s = create(1, 3)
print(s)
```

```python
# another way of creating a queue using list

def create(*n):
    return list(s)


s = create(1, 3)
print(s)
```

```python
# checking if queue is empty

def is_Empty(front=0, rear=-1):
    return rear < front

s = is_Empty(1, 3)
y =  is_Empty()
print(s)
print(y)
```

```python
# another way of checking if queue is empty

def is_empty(queue):
    return len(queue) == 0

print(is_empty([]))          # True
print(is_empty([1, 2, 3]))   # False
```

```python
# enqueue an element

def enqueue(q, el):
    q.append(el)
    return q

s = enqueue([1, 7, 9], 3)
print(s)
```

```python
# code for enqueuing an element with given size

def enqueue(q, el, max_size):
    if len(q) < max_size:
        q.append(el)
        return q
    return 'Queue Full'

print(enqueue([1, 7, 9], 3, 5))
```

```python
# code for dequeuing an element

def dequeue(q):
    if len(q) > 0:
        return q.pop(0)

    return 'Queue Empty'

q = [1, 7, 9]

q = dequeue(q)
print(q)
```

```python
# code for displaying element


def display(q):
    if not q:
        return 'Queue Empty'
    for el in q:
        print(el)
    return q


q = [1, 7, 9]

q = display(q)
print(q)
```

```python
# code for displaying the whole queue element


def display(q):
    if not q:
         return 'Queue Empty'

    return q

q = [1, 7, 9]

q = display(q)
print(q)
```

```python
# class queue implementation

class Queue:

    def __init__(self, size):
        self.items = []
        self.size = size

    def is_Empty(self):
        return len(self.items) == 0

    def is_Full(self):
        return len(self.items) >= self.size

    def enqueue(self, el):
        if self.is_Full():
            return "Queue Full"

        self.items.append(el)

    def dequeue(self):
        if self.is_Empty():
            return "Queue Empty"

        return self.items.pop(0)

    def display(self):
        if self.is_Empty():
            return "Queue Empty"

        return self.items
```

```python
# test of code for class queue implementation

q = Queue(5)
print('Queue Empty?:', is_Empty(q))
print()

enqueue(23)
enqueue(9)
enqueue(41)
enqueue(0)
print('Queue:', display(q))
print('Queue Full?:', is_Empty(q))
print()

enqueue(17)
enqueue(82)
enqueue(35)
print('Queue:', display(q))
print('Queue Full?:', is_Full(q))
print()

dequeue(17)
dequeue(82)
print('Queue:', display(q))
```
