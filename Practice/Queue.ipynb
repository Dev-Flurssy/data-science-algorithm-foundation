```python
# creating a queue using list

def create(*s):
    q = []
    q.extend(s)
    return q


s = create(1, 3)
print(s)
```

```python
# another way of creating a queue using list

def create(*n):
    return list(s)


s = create(1, 3)
print(s)
```

```python
# checking if queue is empty

def is_Empty(front=0, rear=-1):
    return rear < front

s = is_Empty(1, 3)
y =  is_Empty()
print(s)
print(y)
```

```python
# another way of checking if queue is empty

def is_Empty(queue):
    return len(queue) == 0

s = is_Empty(1, 3)
y =  is_Empty(([]))
print(s)
print(y)
```
