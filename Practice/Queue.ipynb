```python
# creating a queue using list

def create(*s):
    q = []
    q.extend(s)
    return q


s = create(1, 3)
print(s)
```

```python
# another way of checking if queue is empty

def is_empty(queue):
    return len(queue) == 0

print(is_empty([]))          # True
print(is_empty([1, 2, 3]))   # False
```

```python
# enqueue an element

def enqueue(q, el):
    q.append(el)
    return q

s = enqueue([1, 7, 9], 3)
print(s)
```

```python
# code for enqueuing an element with given size

def enqueue(q, el, max_size):
    if len(q) < max_size:
        q.append(el)
        return q
    return 'Queue Full'

print(enqueue([1, 7, 9], 3, 5))
```

```python
# code for dequeuing an element

def dequeue(q):
    if len(q) > 0:
        return q.pop(0)

    return 'Queue Empty'

q = [1, 7, 9]

q = dequeue(q)
print(q)
```

```python
# code for displaying element


def display(q):
    if not q:
        return 'Queue Empty'
    for el in q:
        print(el)
    return q


q = [1, 7, 9]

q = display(q)
print(q)
```

```python
# code for displaying the whole queue element


def display(q):
    if not q:
         return 'Queue Empty'

    return q

q = [1, 7, 9]

q = display(q)
print(q)
```

```python
# class queue implementation

class Queue:

    def __init__(self, size):
        self.items = []
        self.size = size

    def is_Empty(self):
        return len(self.items) == 0

    def is_Full(self):
        return len(self.items) >= self.size

    def enqueue(self, el):
        if self.is_Full():
            return "Queue Full"

        self.items.append(el)

    def dequeue(self):
        if self.is_Empty():
            return "Queue Empty"

        return self.items.pop(0)

    def display(self):
        if self.is_Empty():
            return "Queue Empty"

        return self.items
```

```python
# test of code for class queue implementation

q = Queue(5)

print("Queue Empty?:", q.is_Empty())
print()

q.enqueue(23)
q.enqueue(9)
q.enqueue(41)
q.enqueue(0)

print("Queue:", q.display())
print("Queue Full?:", q.is_Full())
print()

q.enqueue(17)
q.enqueue(82)  
q.enqueue(35)  

print("Queue:", q.display())
print("Queue Full?:", q.is_Full())
print()

print("Dequeued:", q.dequeue())
print("Dequeued:", q.dequeue())

print("Queue:", q.display())
```

```python
# code for creating a queue using a dictionary

def create():
    q = {}
    front=0
    rear=-1
    return q, front, rear
```

```python
# code for enqueuing using a dictionary

def enqueue(q, el):
    q[len(q)] = el
    return q

q = {0: 1, 1: 7, 2: 9}
print(enqueue(q, 3))
```

```python
# code for dequeuing using a dictionary

def dequeue(q):
    if not q:
        return 'Queue Empty'

    q.pop(0)
    return q


q = {0: 1, 1:9, 2: 7, 3: 0, 4:5}
print(dequeue(q))
```

```python
# another code for dequeuing using a class

def dequeue(q):
    if not q:
        return 'Queue Empty'
    front = min(q.keys())
    removed = q.pop(front)
    
    print('Dequeued:', removed)
    return q

q = {0: 1, 1:9, 2: 7, 3: 0, 4:5}
print(dequeue(q))
```
