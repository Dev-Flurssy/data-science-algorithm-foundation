```python
graph = {
    'E': ['B', 'G'],
    'G': ['A', 'E'],
    'B': ['A', 'E', 'F'],
    'A': ['B', 'D', 'G'],
    'D': ['A', 'F'],
    'F': ['B', 'C', 'D'],
    'C': ['F', 'H'],
    'H': ['C']
}
```

```python
# Recursive DFS implementation

def dfs(graph, node, visited=None):
    if visited is None:
        visited = set()

    visited.add(node)
    print(node, end=" ")

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)


print('Recursive Depth-First Search:')
dfs(graph, 'A')
```

```python
# DFS implementation

def dfs(graph, start):
    visited = set()
    traversal = []

    def explore(node):
        visited.add(node)
        traversal.append(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                explore(neighbor)

    explore(start)
    return traversal

print(dfs(graph, 'A'))
```

```python
#  Stack DFS implementation

def dfs(graph, start):
    stack = [start]
    visited = set()
    traversal = []

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.add(node)
            traversal.append(node)

            for neighbor in reversed(graph[node]):
                stack.append(neighbor)

    return traversal

print(dfs(graph, 'A'))
```

```python
# BFS implementation

from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


print('Breadth-First Search:')
bfs(graph, 'A')
```
