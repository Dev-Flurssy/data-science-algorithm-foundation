```python
# BFS implementation


from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])
    transversal = []

    while queue:
        node = queue.popleft()
        transversal.append(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return transversal

graph = {
    'E': ['B', 'G'],
    'G': ['A', 'E'],
    'B': ['A', 'E', 'F'],
    'A': ['B', 'D', 'G'],
    'D': ['A', 'F'],
    'F': ['B', 'C', 'D'],
    'C': ['F', 'H'],
    'H': ['C']
}

print(bfs, 'A')
```
