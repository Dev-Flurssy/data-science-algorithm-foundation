```python
# push elements to stack

def push_elements(stack, top):
    count = int(input("How many numbers? "))

    for _ in range(count):
        value = int(input("Enter a number: "))
        top += 1
        stack.append(value)

    return stack, top

s = []
top = -1
datas = push_elements(s, top)
datas
```

```python
# reverse stack implementation

def reverse_stack(s):
    if s:
        for item in reversed(s):
            print(item)
    else:
        print('Stack empty')

s = [10, 20, 30, 40]
reverse_stack(s)
```

```python
# reverse stack list implementation

def reverse_stack_list(s):
    if s:
        return list(reversed(s))
    else:
        print('Stack empty')

s = [10, 20, 30, 40]
print(reverse_stack_list(s))
```

```python
# stack class implementation

class Stack:

    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return "Stack Empty"

    def peek(self):
        if not self.is_empty():
            return self.items[-1]
        return "Stack Empty"

    def display(self):
        if not self.is_empty():
            return self.items
        return "Stack Empty"

    def display_reversed(self):
        if not self.is_empty():
            return self.items[::-1]
        return "Stack Empty"
```

```python
# test of class stack implementation

s = Stack()

s.push(12)
s.push(67)
s.push(39)
s.push(51)

print('Stack:', s.display())
print('Top Element:', s.peek())
print('Revered Stack: ', s.display_reversed())
print('Top Element:', s.peek())
print()

s.pop()
s.pop()
print('Stack is Empty?:', s.is_empty())
print('Stack:', s.display())
print()

s.push(25)
s.push(77)
s.push(19)
print('Stack:', s.display())
```

```python
# another stack pop implementation using dictionary 

def pop(stack):
    if not stack:
        return "Stack Empty"

    top = len(stack) - 1
    value = stack[top]

    del stack[top]

    return value

s = {
    0: 10,
    1: 20,
    2: 30
}

print(pop(s))
print(s)
```

```python
# class stack implementation using dictionary

class Stack:

    def __init__(self, size):
        self.items = {}
        self.size = size

    def is_Empty(self):
        return len(self.items) == 0

    def is_Full(self):
        return len(self.items) >= self.size

    def push(self, item):
        if self.is_Full():
            return "Stack Full"

        self.items[len(self.items)] = item

    def pop(self):
        if self.is_Empty():
            return "Stack Empty"

        return self.items.pop(len(self.items) - 1)

    def peek(self):
        if self.is_Empty():
            return "Stack Empty"

        return self.items[len(self.items) - 1]

    def display(self):
        return self.items
```

```python
# test of class stack implementation using dictionary

s = Stack(3)

s.push(10)
s.push(20)
s.push(30)

print(s.display())
print(s.peek())
print(s.pop())
print(s.display())
```
