```python
# empty stack implementation

def create_stack():
    s = []
    top = -1
    print('Empty stack created!')
    return s, top

create_stack()
```

```python
# check if stack is empty implementation

def isEmpty(stack):
    return len(stack) == 0

s= []
isEmpty(s)
```

```python
# another way of checking if stack is empty implementation


def isEmpty(s):
    return not s

s = [10, 20]
print(isEmpty(s))
```

```python
# push element to stack

def push_element(s, top):
    data = int(input('please enter number: '))
    top += 1
    s.append(data)
    return s, top

s = []
top = -1
datas = push_element(s, top)
datas
```

```python
# push elements to stack

def push_elements(stack, top):
    count = int(input("How many numbers? "))

    for _ in range(count):
        value = int(input("Enter a number: "))
        top += 1
        stack.append(value)

    return stack, top

s = []
top = -1
datas = push_elements(s, top)
datas
```

```python
# pop element in a stack

def pop_element(s, top):
    if top > -1:
        print(s.pop())
        print(s)
    else:
        print('Stack empty')

s = [10, 20, 30]
top = 2

pop_element(s, top)
```

```python
# reverse stack implementation

def reverse_stack(s):
    if s:
        for item in reversed(s):
            print(item)
    else:
        print('Stack empty')

s = [10, 20, 30, 40]
reverse_stack(s)
```

```python
# reverse stack list implementation

def reverse_stack_list(s):
    if s:
        return list(reversed(s))
    else:
        print('Stack empty')

s = [10, 20, 30, 40]
print(reverse_stack_list(s))
```

```python
# Add pythonic stack list implementation 

def reverse_stack_list(s):
    if s:
        return s[::-1]
    else:
        print('Stack empty')

s = [10, 20, 30, 40]
print(reverse_stack_list(s))
```
