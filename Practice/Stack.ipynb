```python
# empty stack implementation

def create_stack():
    s = []
    top = -1
    print('Empty stack created!')
    return s, top

create_stack()
```

```python
# check if stack is empty implementation

def isEmpty(stack):
    return len(stack) == 0

s= []
isEmpty(s)
```

```python
# another way of checking if stack is empty implementation


def isEmpty(s):
    return not s

s = [10, 20]
print(isEmpty(s))
```
